```python
import numpy as np
import matplotlib.pyplot as plt


score=np.random.normal(70,10,100)
plt.hist(score,color="red")
```

```python
#multiple transcation
legit_transcation=[2.99,5.49,8.99,12.55,14.99,19.99,23.45,29.33,34.45,39.45,
                   45.00,49.99,55.55]
fraud_transcation=[50,100,150,200,250,300,350,400,450,500,750,1000,1000,1200,1500,1500,2000,2500,3000,3000]


plt.hist(fraud_transcation,color="blue",bins=10,label="fraud-transaction",alpha=0.5)
plt.hist(legit_transcation,color="green",bins=10,label="legit-transaction",alpha=0.5)
plt.title("legit Vs Fraud Transcation")
plt.grid(True)
```

```python
from matplotlib.lines import lineStyles
marks=[10,12,15,17,25,30,45,38,55,73,80,88,90,45,72,66,23,77,88,89,90,33,71,64,37,47]

plt.axvline(
    x=33,
    label="passing marks",
    color="red",
    linewidth=2,
    linestyle="--"
)
plt.hist(marks,color="skyblue",edgecolor="black")
plt.title("the number of students pass the exam ")
plt.ylabel("no of students")
plt.xlabel("marks ")
```

```python
#boxPlot
data=[7,8,5,6,9,4,10,12,15]
plt.boxplot(data,vert=False,showmeans=True)
plt.grid(True)
```

```python
group1=np.random.normal(50,10,100)
group2=np.random.normal(60,15,100)

plt.boxplot([group1,group2],tick_labels=["group1","group2"])
plt.show()
```

```python
#stack plot
days=["mon","tue","wed","thur","fri","sat","sun"]
direct=[50,60,70,80,90,100,110]
organic=[30,40,50,55,60,70,80]
social=[20,25,30,35,40,50,60]

# plt.pie([direct[0],organic[0],social[0]],labels=["Direct","organic","social"], autopct="%1.1f%%")

plt.stackplot(days,direct,organic,social,labels=["diect","organic","social"])
plt.show()
plt.legend()
```
