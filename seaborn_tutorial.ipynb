```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme()
```

```python
sns.get_dataset_names()
```

```python
tips=sns.load_dataset("tips")
print(tips.head())
```

```python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="smoker",
    size="size",
    style="smoker"
)
```

```python
tips=sns.load_dataset("tips")
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="time"
)
```

```python
flight=sns.load_dataset("flights")
sns.lineplot(
    data=flight,
    x="year",
    y="passengers"
)
```

```python
sns.barplot(
    data=tips,
    x="day",
    y="tip",
    hue="sex"
)
```

```python
sns.boxplot(
    data=tips,
    x="day",
    y="tip"
)
```

```python
penguins=sns.load_dataset("penguins")
penguins.head()
sns.histplot(
    data=penguins,
    x="body_mass_g",
    bins=40
)
```

```python
flights=sns.load_dataset("flights")
flight_pivot=flights.pivot(index="month",columns="year",values="passengers")


sns.heatmap(
    flight_pivot,
    cmap="coolwarm",
    fmt="d"
)
```
